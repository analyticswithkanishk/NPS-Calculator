```python
#NPS -- Net Promoter Score 
#NPS is a real-world business metric used by companies like Apple, Netflix, Amazon. You'll learn a ton building this. 

#Ask users: "On a scale of 0–10, how likely are you to recommend us?"
#Promoters = score 9–10 (love you!)
#Passives = score 7–8 (neutral)
#Detractors = score 0–6 (unhappy)
#NPS = % Promoters − % Detractors (ranges from -100 to +100)

```

```python
import numpy as np
```

```python
np.random.seed(42) 
# np is the alias (nickname) for the NumPy library
# .random is used to generate random numbers
# .seed() locks the random numbers so they stay SAME every time we run
# 42 is just a programmer's choice — any number works (1, 100, 999, etc.)
```

```python
ratings=np.random.randint(0,11,size=1000)
# ratings is a variable that stores all the scores
#np.random is go to the random section
# randint generates random whole numbers
# 0 is the start, 11 is the stop (10 is the max, 11 is excluded)
# size=1000 means we are generating 1000 random ratings
```

```python
ratings
```

```python
# Promoters  --- score >= 9   (9 and 10)
# Passives   --- score == 7 or score == 8   (7 and 8)
# Detractors --- score <= 6   (0 to 6)
```

```python
promoters  = ratings[ratings >= 9]
# filter all scores >= 9 (score 9 and 10) → promoters

detractors = ratings[ratings <= 6]
# filter all scores <= 6 (score 0 to 6) → detractors

passive    = ratings[(ratings >= 7) & (ratings <= 8)]
# filter all scores between 7 and 8 → passives
# & means BOTH conditions must be true at the same time
```

```python
print(promoters)
```

```python
print(passive)
```

```python
print(detractors)
```

```python
len_promoters  = len(promoters)
# counting total number of promoters

len_detractors = len(detractors)
# counting total number of detractors

total_len      = len(ratings)
# counting total ratings (1000)
```

```python
promoters_percent = len_promoters / total_len * 100
# calculating % of promoters out of total ratings

detractor_percent = len_detractors / total_len * 100
# calculating % of detractors out of total ratings
```

```python
print(promoters_percent)
```

```python
print(detractor_percent)
```

```python
nps = promoters_percent - detractor_percent
# NPS formula = % promoters - % detractors
# result is the final Net Promoter Score
```

```python
print(nps) 
#print(f"Net Promoter Score: {nps:.2f}")
# Output → Net Promoter Score: -48.00 (.2f means only 2 digit float number)
```

```python
if nps > 0:
    print(f"NPS: {nps:.2f} Company is doing good job and moving towards growth! ")
elif nps < 0:
    print(f"NPS: {nps:.2f} Company is moving towards decline! ")
else:
    print(f"NPS: {nps:.2f} Company is neither in growth nor decline! ")
```
